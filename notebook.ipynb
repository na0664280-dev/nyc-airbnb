```python
import plotly.express as px

# 1) متوسط السعر حسب المنطقة
a = df.groupby("neighbourhood_group")["price"].mean().sort_values().reset_index()
px.bar(a, x="neighbourhood_group", y="price")
```

```python
# 2) السعر حسب نوع الغرفة
b = df.groupby("room_type")["price"].median().reset_index()
px.bar(b, x="room_type", y="price")
```

```python
# 3) عدد الإعلانات حسب المنطقة
c = df["neighbourhood_group"].value_counts().reset_index()
px.bar(c, x="neighbourhood_group", y="count")
```
